LCEL (LangChain Expression Language) is a declarative syntax that makes it incredibly easy to compose complex AI chains in LangChain. 

At the heart of LCEL is the Runnable protocol. Every component (such as an LLM, a prompt template, or an output parser) is treated as a "Runnable." When you use the pipe operator (|), LangChain takes the output of the component on the left and passes it as the input to the component on the right.

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_openai import ChatOpenAI

# 1. Define your components
prompt = ChatPromptTemplate.from_template("Tell me a short joke about {topic}")
model = ChatOpenAI(model="gpt-4o")
output_parser = StrOutputParser()

# 2. Build the LCEL chain using the | operator
chain = prompt | model | output_parser

# 3. Invoke the chain
result = chain.invoke({"topic": "AI"})
print(result)
```
